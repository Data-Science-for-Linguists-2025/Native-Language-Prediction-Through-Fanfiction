```python
import spacy
import stanza
import pickle
import dask
from dask.distributed import Client, LocalCluster
from bs4 import BeautifulSoup as beau
```

```python
spcy = spacy.load("en_core_web_sm")
stnz = stanza.Pipeline('en', download_method=None)
```

```python
fics_df = pickle.load(open('fics_df_manual.pkl', 'rb'))
```

```python
client = Client(processes=True, n_workers=4, threads_per_worker=200)
```

```python
df = fics_df[['L1', 'work']][:5]
df['word_y'] = df.work.map(lambda x : [y.text for y in [spcy(beau(z).text) for z in x]])
df['word_z'] = df.work.map(lambda x : [w.text for w in [s for s in stnz(x).sentences]])
df['pos_y'] = df.work.map(lambda x : [y.pos_ for y in spcy(x)])
df['pos_z'] = df.work.map(lambda x : [w.pos for w in [s.words for s in stnz(x).sentences]])
```
